```python
numeros = [1, 2, 3, 4]
pares = []
i = 0  # Empezamos en el índice 0 (la primera posición de la lista)

while i < len(numeros):  # Mientras 'i' no se pase del tamaño de la lista
    if numeros[i] % 2 == 0:  # Mira el número que hay en la posición 'i'
        pares.append(numeros[i])
    i = i + 1  # Avanzamos a la siguiente posición (fuera del if)

print(pares)  # Imprime: [2, 4]
```
